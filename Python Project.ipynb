```python
import json
import os
from datetime import datetime

DATA_FILE = "library_data.json"

# Load existing books or create sample data
def load_data():
    if not os.path.exists(DATA_FILE):
        default_books = [
            {"id": "B101", "title": "Python Programming", "author": "Guido van Rossum", "status": "Available", "borrower": "", "issue_date": ""},
            {"id": "B102", "title": "Data Structures", "author": "Narasimha Karumanchi", "status": "Available", "borrower": "", "issue_date": ""}
        ]
        save_data(default_books)
        return default_books
    with open(DATA_FILE, "r") as f:
        return json.load(f)

def save_data(books):
    with open(DATA_FILE, "w") as f:
        json.dump(books, f, indent=4)

def view_books(books):
    print("\n--- CURRENT LIBRARY CATALOG ---")
    if not books:
        print("No books in the library.")
        return
    for book in books:
        print(f"ID: {book['id']} | Title: {book['title']} | Author: {book['author']} | Status: {book['status']}")
        if book['status'] == 'Issued':
            print(f"   -> Borrowed by: {book['borrower']} on {book['issue_date']}")

def add_book(books):
    book_id = input("Enter Book ID: ").strip()
    for book in books:
        if book['id'] == book_id:
            print("Error: Book ID already exists!")
            return
    title = input("Enter Book Title: ").strip()
    author = input("Enter Author Name: ").strip()

    new_book = {
        "id": book_id,
        "title": title,
        "author": author,
        "status": "Available",
        "borrower": "",
        "issue_date": ""
    }
    books.append(new_book)
    save_data(books)
    print(f"Book '{title}' added successfully!")

def search_book(books):
    query = input("Enter Book Title or ID to search: ").strip().lower()
    found = False
    for book in books:
        if query in book['id'].lower() or query in book['title'].lower():
            print(f"\nFound -> ID: {book['id']} | Title: {book['title']} | Author: {book['author']} | Status: {book['status']}")
            found = True
    if not found:
        print("No matching books found.")

def issue_book(books):
    book_id = input("Enter Book ID to issue: ").strip()
    for book in books:
        if book['id'] == book_id:
            if book['status'] == 'Issued':
                print("Error: Book is already issued!")
                return
            borrower = input("Enter Student Name: ").strip()
            book['status'] = 'Issued'
            book['borrower'] = borrower
            book['issue_date'] = datetime.now().strftime("%Y-%m-%d")
            save_data(books)
            print(f"Book '{book['title']}' successfully issued to {borrower}!")
            return
    print("Error: Book ID not found.")

def return_book(books):
    book_id = input("Enter Book ID being returned: ").strip()
    for book in books:
        if book['id'] == book_id:
            if book['status'] == 'Available':
                print("Error: This book was not issued!")
                return

            issue_dt = datetime.strptime(book['issue_date'], "%Y-%m-%d")
            current_dt = datetime.now()
            days_diff = (current_dt - issue_dt).days
            due_days = 7  # Standard loan period

            print(f"Issued on: {book['issue_date']} (Duration elapsed: {days_diff} days)")

            if days_diff > due_days:
                overdue_days = days_diff - due_days
                fine = overdue_days * 10  # ₹10 fine per overdue day
                print(f"Days overdue: {overdue_days} days. Late fine calculated: ₹{fine}")
            else:
                print("Returned on time. No fine applicable.")

            book['status'] = 'Available'
            book['borrower'] = ""
            book['issue_date'] = ""
            save_data(books)
            print("Book returned successfully.")
            return
    print("Error: Book ID not found.")

def main():
    books = load_data()
    while True:
        print("\n=== LIBRARY MANAGEMENT SYSTEM ===")
        print("1. View All Books")
        print("2. Add a New Book")
        print("3. Search for a Book")
        print("4. Issue a Book")
        print("5. Return a Book")
        print("6. Exit")

        choice = input("Enter your choice: ").strip()

        if choice == '1':
            view_books(books)
        elif choice == '2':
            add_book(books)
        elif choice == '3':
            search_book(books)
        elif choice == '4':
            issue_book(books)
        elif choice == '5':
            return_book(books)
        elif choice == '6':
            print("Exiting program. Goodbye!")
            break
        else:
            print("Invalid choice! Please enter a number between 1 and 6.")

if __name__ == "__main__":
    main()
```
